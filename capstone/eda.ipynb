# PYSPARK

```python
from pyspark.sql import SparkSession
from pathlib import Path
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder \
    .appName("Capstone Dataset Inspection") \
    .getOrCreate()
```

```python
data_dir = Path("capstone dataset")
october_data_path = str(data_dir / "2019-Oct.csv")
november_data_path = str(data_dir / "2019-Nov.csv")
```

```python
df_october = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(october_data_path)
)

df_november = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(november_data_path)
)
```

## October dataset

```python
from pyspark.sql import functions as F

df_october.show(5)
```

### check data schema

```python
df_october.printSchema()
```

### check null values

```python
df_october.select(
    [F.count(F.when(F.col(column).isNull(), column)).alias(column)
     for column in df_october.columns]
).show()
```

## November dataset

```python
df_november.show(5)
```

### check dataset schema

```python
df_november.printSchema()
```

### check null values

```python
df_november.select(
    [F.count(F.when(F.col(column).isNull(), column)).alias(column)
     for column in df_november.columns]
).show()
```

## overview -- side-by-side

```python
print("october null values")
df_october.select(
    [F.count(F.when(F.col(column).isNull(), column)).alias(column)
     for column in df_october.columns]
).show()

print("november null values")
df_november.select(
    [F.count(F.when(F.col(column).isNull(), column)).alias(column)
     for column in df_november.columns]
).show()
```
